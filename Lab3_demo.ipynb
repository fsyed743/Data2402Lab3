## Step 3
Finds the two stats required across all patients: average BMI and the busiest month.


**average_bmi:** Loops through every exam, calls each object's own calculate_bmi(), and averages the results.

```python
avg = average_bmi(exams)
print(f"{avg:.2f}")
```


**busiest_month:** Creates a {month: count} dictionary , then uses max(month_counts, key=month_counts.get) to find the month with the highest count, not the highest month number, which is what a plain max(month_counts) would give.

```python
month = busiest_month(exams)
print(month)
```

## Normal Case
**Expected:** average BMI ≈ 21.60, busiest month = 6 (June).

```python
exams_normal = load_exams("patient_data.csv")
avg_normal = average_bmi(exams_normal)
month_normal = busiest_month(exams_normal)
print("Rows:", len(exams_normal))
print(f"Average BMI: {avg_normal:.2f}")
print("Busiest month:", month_normal)
```

## Special case (One Patient)
**Expected:** average_bmi on a one item list just returns that patient's own BMI, and busiest_month returns that patient's own month.

```python
single_exam = PatientExam(1, "3/5/2020", "John Smith", 70, 1.75)
single_list = [single_exam]
print(single_exam.calculate_bmi())
print(average_bmi(single_list))
print(busiest_month(single_list))
```
